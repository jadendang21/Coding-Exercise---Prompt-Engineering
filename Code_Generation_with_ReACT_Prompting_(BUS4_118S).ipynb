```python
from google.colab import userdata
from openai import OpenAI

api_key = userdata.get("OPENAI_API_KEY")

client = OpenAI(api_key=api_key)

def ask_ai(prompt):
    response = client.responses.create(
        model="gpt-5.6",
        input=prompt
    )
    return response.output_text
```

```python
react_prompt = """
You are an expert Python programmer.

Your task is to create a Python program that analyzes a list of sales values.

Requirements:
- Input: a Python list of numeric sales values.
- Calculate total sales.
- Calculate average sales.
- Find the highest sale.
- Find the lowest sale.
- Count how many sales are above average.
- Handle an empty list without crashing.
- Use only standard Python libraries.
- Format monetary results to 2 decimal places.

Follow this ReACT-style process:

PLAN:
Briefly describe the programming approach.

ACT:
Generate the Python code.

RUN:
Explain what test data should be used to run the program.

OBSERVE:
Evaluate whether the expected output satisfies all requirements.

FIX:
If a possible error or edge case exists, revise the code.

FINAL CODE:
Provide the final corrected Python code only.
"""

result = ask_ai(react_prompt)

print(result)
```

```python
def analyze_sales(sales):
    if not sales:
        print("No sales data available.")
        return

    total = sum(sales)
    average = total / len(sales)
    highest = max(sales)
    lowest = min(sales)
    above_average = sum(1 for sale in sales if sale > average)

    print(f"Total Sales: ${total:,.2f}")
    print(f"Average Sales: ${average:,.2f}")
    print(f"Highest Sale: ${highest:,.2f}")
    print(f"Lowest Sale: ${lowest:,.2f}")
    print(f"Sales Above Average: {above_average}")


sales_data = [1250, 900, 1450, 700, 1800]

analyze_sales(sales_data)
```

```python
analyze_sales([])
```
