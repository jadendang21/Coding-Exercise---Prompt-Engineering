```python
from google.colab import userdata
from openai import OpenAI

api_key = userdata.get("OPENAI_API_KEY")

client = OpenAI(api_key=api_key)

def ask_ai(prompt):
    response = client.responses.create(
        model="gpt-5.6",
        input=prompt
    )
    return response.output_text
```

```python
source_text = """
Artificial intelligence allows computers to perform tasks that normally
require human intelligence. AI systems can recognize patterns, analyze
large amounts of data, make predictions, understand language, and automate
repetitive tasks. Businesses use AI for customer service, fraud detection,
recommendation systems, forecasting, and process automation.
"""
```

```python
summary_prompt = f"""
Summarize the following passage.

Requirements:
- Keep the summary under 100 words.
- Write one paragraph.
- Focus on the main ideas.

Text:
{source_text}
"""

original_summary = ask_ai(summary_prompt)

print("ORIGINAL SUMMARY")
print(original_summary)
```

```python
reflection_prompt = f"""
You are reviewing a summary you previously created.

Evaluate the summary using these criteria:

1. Accuracy - Does it correctly represent the source text?
2. Clarity - Is it easy to understand?
3. Relevance - Does it focus only on the most important information?
4. Audience - Is it appropriate for a college student?
5. Length - Is it under 75 words?
6. Formatting - Is it written as one concise paragraph?

First, identify weaknesses in the original summary.

Then rewrite the summary so that it satisfies all six requirements.

SOURCE TEXT:
{source_text}

ORIGINAL SUMMARY:
{original_summary}

Return exactly:

CRITIQUE:
[Your critique]

IMPROVED SUMMARY:
[Your revised summary]
"""

improved_summary = ask_ai(reflection_prompt)

print(improved_summary)
```

```python
### Self-Reflection and Improvement

#**Original Output:**
#The AI first created a summary of the source text.

#**Reflection Criteria:**
#The summary was evaluated for accuracy, clarity, relevance, audience, length, and formatting.

#**Improvement:**
#The AI identified weaknesses in the original summary and created a revised version that was shorter, clearer, and more focused.

#**Result:**
#The improved summary better followed the prompt requirements.
```
