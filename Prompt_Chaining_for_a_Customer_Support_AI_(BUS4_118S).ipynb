```python
from google.colab import userdata
from openai import OpenAI

api_key = userdata.get("OPENAI_API_KEY")

client = OpenAI(api_key=api_key)

print("API key loaded successfully")
```

```python
customer_message = """
I ordered a laptop three days ago and paid for overnight shipping,
but the tracking page still says the package has not shipped.
I need the laptop for work tomorrow.
"""
```

```python
def ask_ai(prompt):
    response = client.responses.create(
        model="gpt-5.6",
        input=prompt
    )
    return response.output_text
```

```python
prompt1 = f"""
You are a customer support assistant.

Classify the customer's problem into ONE category:
Shipping, Billing, Product Issue, Refund, Account, or Other.

Also assign urgency as Low, Medium, or High.

Return exactly:
Category:
Urgency:
Reason:

Customer message:
{customer_message}
"""

classification = ask_ai(prompt1)

print("STEP 1 - CLASSIFICATION")
print(classification)
```

```python
prompt2 = f"""
You are a customer support assistant.

Use the customer's message AND the previous classification.

Determine what additional information is needed before solving the problem.

Ask no more than 3 questions.
Do not request passwords, credit card numbers, or other sensitive information.

Customer message:
{customer_message}

Previous classification:
{classification}

Return:
Missing Information:
Questions:
"""

missing_info = ask_ai(prompt2)

print("STEP 2 - INFORMATION NEEDED")
print(missing_info)
```

```python
prompt3 = f"""
You are a professional customer support representative.

Use all previous information to suggest an appropriate solution.

Requirements:
- Be polite and concise.
- Do not promise refunds or compensation without authorization.
- Give actionable next steps.
- Keep the response under 120 words.

Customer message:
{customer_message}

Classification:
{classification}

Information analysis:
{missing_info}

Return:
Recommended Solution:
Customer Response:
"""

solution = ask_ai(prompt3)

print("STEP 3 - SOLUTION")
print(solution)
```

```python
prompt4 = f"""
You are a customer support escalation assistant.

Review the entire support case.

Escalate if:
- The issue is high urgency,
- The customer may miss an important deadline,
- The normal solution may not resolve the issue,
- A refund or manager approval may be needed.

Return exactly:

Escalate: Yes or No
Reason:
Escalation Team:

Customer message:
{customer_message}

Classification:
{classification}

Missing information:
{missing_info}

Proposed solution:
{solution}
"""

escalation = ask_ai(prompt4)

print("STEP 4 - ESCALATION")
print(escalation)
```

```python
### Testing and Iteration

#**Initial Test:**
#The first classification prompt only identified the type of customer issue.

#**Improvement:**
#I improved the prompt by adding urgency levels, clear categories, and a required output format.

#**Result:**
#The revised prompt produced a more consistent response that could be passed into the next steps of the prompt chain.
```

```python
### Prompt Chain Explanation

#Step 1 classifies the customer issue and urgency.

#Step 2 uses the Step 1 classification to determine what information is missing.

#Step 3 uses the customer message, classification, and missing information to propose a solution.

#Step 4 reviews all previous outputs and decides whether the issue should be escalated.
```
